```python
import numpy as np, pandas as pd

HIST_PATH = r"C:/Users/anasz/Desktop/uni bitch/FYP shi/data/Traffic_Volume_Counts_(Historical)_20260926.csv"      
rawh = pd.read_csv(HIST_PATH)
assert rawh.shape[1] == 31, f"Wrong file: {rawh.shape}"
print("raw:", rawh.shape)

hour_cols = [f"h{i:02d}" for i in range(24)]
hh = rawh.copy()
hh.columns = list(hh.columns[:7]) + hour_cols

for c in hour_cols:
    hh[c] = pd.to_numeric(hh[c].astype(str).str.replace(",", "", regex=False).str.strip(), errors="coerce")
junk = int((rawh.iloc[:, 7:].notna().values & hh[hour_cols].isna().values).sum())
print("Hour cells that were non-null but not numeric:", junk)

hh["Date"] = pd.to_datetime(hh["Date"], format="%m/%d/%Y", errors="coerce")
print("Unparsed dates:", int(hh["Date"].isna().sum()))

# Direction: base direction plus road type
dir_raw = hh["Direction"].astype(str).str.upper().str.strip()
mm = dir_raw.str.extract(r"^(NB|SB|EB|WB)\s*-?\s*(MAIN|SER|SR|MR)?$")
print("Directions not parsed:", int(mm[0].isna().sum()))
hh["direction"], hh["dir_suffix"] = mm[0], mm[1]
hh["road_type"] = hh["dir_suffix"].map({"MAIN": "main", "MR": "main",
                                        "SER": "service", "SR": "service"}).fillna("unspecified")

# Quarantine
key_w = ["SegmentID", "direction", "road_type", "Date"]
hh["q_reason"] = pd.Series(pd.NA, index=hh.index, dtype="object")
hh.loc[hh["direction"].isna(), "q_reason"] = "bad_direction"
rem = hh["q_reason"].isna()
dup = hh.loc[rem].duplicated(key_w, keep=False)
hh.loc[dup[dup].index, "q_reason"] = "duplicate_key"
print(hh["q_reason"].value_counts(dropna=False))
print("For comparison, raw duplicate-key rows:",
      int(rawh.duplicated(["SegmentID", "Direction", "Date"], keep=False).sum()), "(earlier: 350)")
```

```python
base = hh[hh["q_reason"].isna()].copy()
long = base.melt(id_vars=["ID", "SegmentID", "direction", "road_type", "Date"],
                 value_vars=hour_cols, var_name="hr_col", value_name="Vol")
long["hr"] = long["hr_col"].str[1:].astype(int)
long["hour_start"] = long["Date"] + pd.to_timedelta(long["hr"], unit="h")
n_cells = len(long)
n_nonnull = int(long["Vol"].notna().sum())
print("Hour cells:", n_cells, "| non-null:", n_nonnull)

skey = ["SegmentID", "direction", "road_type"]
long = (long.dropna(subset=["Vol"]).sort_values(skey + ["hour_start"]).reset_index(drop=True))

typ = (long[long["Vol"] > 0].groupby(skey + ["hr"])["Vol"].median().rename("typ").reset_index())
long = long.merge(typ, on=skey + ["hr"], how="left")

ALWAYS_H, EXPECTED_MIN = 2, 8
is0 = long["Vol"].eq(0)
new_run = ((is0 != is0.shift())
           | (long[skey] != long[skey].shift()).any(axis=1)
           | (long["hour_start"].diff() > pd.Timedelta(hours=1)))
run_len = is0.groupby(new_run.cumsum()).transform("size").where(is0, 0)
expected = run_len * long["typ"].fillna(0)
outage = is0 & ((run_len >= ALWAYS_H) | (expected >= EXPECTED_MIN))

print("Zero cells:", int(is0.sum()))
print("  missing, run >= 2 h:", int((is0 & (run_len >= ALWAYS_H)).sum()))
print("  missing, shorter run but expected count >= 8:", int((outage & (run_len < ALWAYS_H)).sum()))
print("  kept as real zeros:", int((is0 & ~outage).sum()))
print("  typical volume (hour median) of the kept zeros:")
print(long.loc[is0 & ~outage, "typ"].describe().round(1))

long["zero_to_missing"] = outage
hourly_h = long[~outage].rename(columns={"Vol": "volume"}).copy()
assert len(hourly_h) == len(long) - int(outage.sum())
print("Hourly rows after zero rule:", len(hourly_h))
```

```python
hourly_h["date"] = hourly_h["hour_start"].dt.normalize()
hourly_h["hours_in_day"] = (hourly_h.groupby(["SegmentID", "direction", "road_type", "date"])["volume"]
                            .transform("size"))
hourly_h["partial_day"] = hourly_h["hours_in_day"] < 24
hourly_h["covid"] = hourly_h["hour_start"] >= "2020-03-01"
hourly_h["sparse_year"] = hourly_h["hour_start"].dt.year.isin([2013, 2018, 2021])
tz = hourly_h["hour_start"].dt.tz_localize("America/New_York", ambiguous="NaT", nonexistent="NaT")
hourly_h["hour_start_tz"] = tz
hourly_h["dst_issue"] = tz.isna()
print("DST-affected hours:", int(hourly_h["dst_issue"].sum()),
      "| partial-day hours:", int(hourly_h["partial_day"].sum()))
print(hourly_h.groupby(hourly_h["hour_start"].dt.year).size())
```

```python
import numpy as np, pandas as pd, geopandas as gpd
from pyproj import Transformer

# 1. LION
LION_PATH = r"C:/Users/anasz/Desktop/uni bitch/FYP shi/data/lion/lion.gdb"        # the same .gdb folder or .shp file you used before
lion = (gpd.read_file(LION_PATH, layer="lion") if LION_PATH.lower().endswith(".gdb")
        else gpd.read_file(LION_PATH))
print(lion.shape, lion.crs)        # expect (243237, 129) and EPSG:2263

# 2. Helpers and lookup tables
def to_int_series(s):
    return pd.to_numeric(s.astype(str).str.replace(",", "", regex=False).str.strip(), errors="coerce")

def id_to_row(col_name):
    v = to_int_series(lion[col_name])
    m = (v.notna() & (v % 1 == 0)).values
    s = pd.Series(lion.index[m], index=v[m].astype("int64").values)
    return s[~s.index.duplicated()]

seg_row, leg_row = id_to_row("SegmentID"), id_to_row("LegacyID")

def norm2(s):
    s = s.astype(str).str.upper().str.replace(r"[^A-Z0-9 ]", " ", regex=True)
    subs = [(r"\bE\b", "EAST"), (r"\bW\b", "WEST"), (r"\bN\b", "NORTH"), (r"\bS\b", "SOUTH"),
            (r"\bAV\b|\bAVE\b", "AVENUE"), (r"\bST\b", "STREET"), (r"\bDR\b", "DRIVE"),
            (r"\bTER\b", "TERRACE"), (r"\bRD\b", "ROAD"), (r"\bBLVD\b", "BOULEVARD"),
            (r"\bPKWY\b", "PARKWAY"), (r"\bBR\b|\bBRDG\b", "BRIDGE"), (r"\bFT\b", "FORT"),
            (r"\bMT\b", "MOUNT"), (r"\bGDNS\b", "GARDENS")]
    for pat, rep in subs:
        s = s.str.replace(pat, rep, regex=True)
    return s.str.replace(r"\s+", " ", regex=True).str.strip()

# 3. Automated segments and distance from each count point to its candidate LION lines
AUTO_PATH = r"C:/Users/anasz/Desktop/uni bitch/FYP shi/data/Automated_Traffic_Volume_Counts_20260928.csv"
raw_a = pd.read_csv(AUTO_PATH, usecols=["SegmentID", "Boro", "WktGeom", "street"])
segs = raw_a.drop_duplicates("SegmentID")[["SegmentID", "Boro", "WktGeom"]].reset_index(drop=True)
pts = gpd.GeoSeries.from_wkt(segs["WktGeom"], crs="EPSG:2263")

def cand_dist(row_map):
    rows = segs["SegmentID"].map(row_map)
    has = rows.notna()
    dist = pd.Series(np.nan, index=segs.index)
    if has.any():
        lines = gpd.GeoSeries(lion.geometry.loc[rows[has].astype(int).values].values,
                              index=segs.index[has], crs="EPSG:2263")
        dist[has] = pts[has].distance(lines)
    return rows, dist

segs["seg_row"], segs["seg_dist"] = cand_dist(seg_row)
segs["leg_row"], segs["leg_dist"] = cand_dist(leg_row)

MAXD = 100
seg_ok, leg_ok = segs["seg_dist"] <= MAXD, segs["leg_dist"] <= MAXD
use_seg = seg_ok & (~leg_ok | (segs["seg_dist"] <= segs["leg_dist"]))
use_leg = leg_ok & ~use_seg
segs["lion_key_used"] = np.select([use_seg, use_leg], ["SegmentID", "LegacyID"], default="none")
segs["lion_row"] = np.where(use_seg, segs["seg_row"], np.where(use_leg, segs["leg_row"], np.nan))
segs["key_conflict"] = (segs["seg_row"].notna() & segs["leg_row"].notna()
                        & (segs["seg_row"] != segs["leg_row"]))
print(segs["lion_key_used"].value_counts())    # expect SegmentID 2892, LegacyID 335, none 164

# 4. Spatial fallback for the unmatched segments, with the improved name check
none = segs[segs["lion_key_used"] == "none"]
npts = gpd.GeoDataFrame({"SegmentID": none["SegmentID"].values},
                        geometry=pts[none.index].values, crs="EPSG:2263")
cand = lion[["Street", "geometry"]].copy()
cand["lion_row"] = cand.index
near = gpd.sjoin_nearest(npts, cand, how="left", distance_col="d_ft")
near = near.sort_values("d_ft").drop_duplicates("SegmentID")

rec_names = (raw_a[raw_a["SegmentID"].isin(near["SegmentID"])]
             .assign(n=lambda x: norm2(x["street"]))
             .groupby("SegmentID")["n"].agg(lambda s: set(s)))
lion_n = norm2(near["Street"])
near["name_ok2"] = [any((l in r) or (r in l) for r in rec_names.get(s, set()))
                    for s, l in zip(near["SegmentID"], lion_n)]
print("Spatial candidates passing the name check:", int(near["name_ok2"].sum()), "of", len(near))   # expect 111 of 164

rowmap = near.loc[near["name_ok2"]].set_index("SegmentID")["lion_row"].astype(int)
segs2 = segs.copy()
hit = segs2["SegmentID"].isin(rowmap.index) & segs2["lion_row"].isna()
segs2.loc[hit, "lion_row"] = segs2.loc[hit, "SegmentID"].map(rowmap)
segs2.loc[hit, "lion_key_used"] = "spatial"
print(segs2["lion_key_used"].value_counts())    # expect SegmentID 2892, LegacyID 335, spatial 111, none 53
```

```python
from pyproj import Transformer
for nm_ in ["lion", "to_int_series", "seg_row", "leg_row", "segs2"]:
    assert nm_ in globals(), f"missing {nm_}: rerun the automated LION blocks first"

def norm3(s):
    s = s.astype(str).str.upper().str.replace(r"[^A-Z0-9 ]", " ", regex=True)
    s = s.str.replace(r"\b(\d+)(ST|ND|RD|TH)\b", r"\1", regex=True)
    subs = [(r"\bE\b", "EAST"), (r"\bW\b", "WEST"), (r"\bN\b", "NORTH"), (r"\bS\b", "SOUTH"),
            (r"\bAV\b|\bAVE\b", "AVENUE"), (r"\bST\b", "STREET"), (r"\bDR\b", "DRIVE"),
            (r"\bTER\b", "TERRACE"), (r"\bRD\b", "ROAD"), (r"\bBLVD\b", "BOULEVARD"),
            (r"\bPKWY\b", "PARKWAY"), (r"\bBR\b|\bBRDG\b", "BRIDGE"), (r"\bFT\b", "FORT"),
            (r"\bMT\b", "MOUNT"), (r"\bGDNS\b", "GARDENS"), (r"\bHWY\b", "HIGHWAY"),
            (r"\bEXPY\b", "EXPRESSWAY"), (r"\bPL\b", "PLACE"), (r"\bCT\b", "COURT"),
            (r"\bSQ\b", "SQUARE"), (r"\bTPKE\b", "TURNPIKE")]
    for pat, rep in subs:
        s = s.str.replace(pat, rep, regex=True)
    return s.str.replace(r"\s+", " ", regex=True).str.strip()

hh["road_n"] = norm3(hh["Roadway Name"])
street_n = norm3(lion["Street"])
names_by_seg = hh.groupby("SegmentID")["road_n"].agg(set)

def agrees(sid, row):
    if pd.isna(row):
        return False
    c = street_n.loc[int(row)]
    return any((c in n) or (n in c) for n in names_by_seg[sid])

hist_segs = hh.drop_duplicates("SegmentID")[["SegmentID"]].reset_index(drop=True)
a_map = segs2.set_index("SegmentID")["lion_row"].dropna()
hist_segs["a_row"] = hist_segs["SegmentID"].map(a_map)
hist_segs["seg_row"] = hist_segs["SegmentID"].map(seg_row)
hist_segs["leg_row"] = hist_segs["SegmentID"].map(leg_row)

def pick(r):
    sid = r["SegmentID"]
    if pd.notna(r["a_row"]):
        return r["a_row"], "automated_validated"
    if pd.notna(r["seg_row"]):
        return r["seg_row"], "SegmentID"
    if pd.notna(r["leg_row"]) and agrees(sid, r["leg_row"]):
        return r["leg_row"], "LegacyID"
    return np.nan, "none"

res = hist_segs.apply(pick, axis=1, result_type="expand")
hist_segs["lion_row"], hist_segs["lion_key_used"] = res[0], res[1]
hist_segs["key_conflict"] = (hist_segs["seg_row"].notna() & hist_segs["leg_row"].notna()
                             & (hist_segs["seg_row"] != hist_segs["leg_row"]))
hist_segs["lion_street_n"] = hist_segs["lion_row"].map(
    lambda r: street_n.loc[int(r)] if pd.notna(r) else np.nan)

print(hist_segs["lion_key_used"].value_counts())
print("Segments matched by both keys to different LION rows:", int(hist_segs["key_conflict"].sum()))

# Row-level name check (flags eras where a SegmentID points to a different street)
rf = hh[key_w + ["road_n"]].drop_duplicates(key_w).merge(
    hist_segs[["SegmentID", "lion_street_n"]], on="SegmentID", how="left")
rf["name_mismatch"] = [isinstance(c, str) and c != "" and not ((c in n) or (n in c))
                       for c, n in zip(rf["lion_street_n"], rf["road_n"])]
print("Wide rows flagged name_mismatch:", int(rf["name_mismatch"].sum()), "of", len(rf))
print(rf.merge(hist_segs[["SegmentID", "lion_key_used"]], on="SegmentID")
        .groupby("lion_key_used")["name_mismatch"].agg(["sum", "size"]))

# LION attributes for matched segments (same cleaning as the automated table)
def blank_to_nan(s):
    s = s.astype(str).str.strip()
    return s.where(~s.isin(["", "nan", "None"]), np.nan)

attrs_h = ["POSTED_SPEED", "Number_Travel_Lanes", "Number_Park_Lanes", "Number_Total_Lanes",
           "StreetWidth_Min", "StreetWidth_Max", "SHAPE_Length", "TrafDir", "TRUCK_ROUTE_TYPE",
           "Snow_Priority", "BikeLane", "LBoro", "RBoro"]
mk = hist_segs["lion_row"].notna()
la = lion.loc[hist_segs.loc[mk, "lion_row"].astype(int).values, attrs_h + ["geometry"]].copy()
la.index = hist_segs.index[mk]
for c in ["POSTED_SPEED", "Number_Travel_Lanes", "Number_Park_Lanes", "Number_Total_Lanes",
          "StreetWidth_Min", "StreetWidth_Max", "TRUCK_ROUTE_TYPE", "BikeLane", "LBoro", "RBoro"]:
    la[c] = pd.to_numeric(blank_to_nan(la[c]), errors="coerce")
for c in ["Number_Park_Lanes", "TRUCK_ROUTE_TYPE", "BikeLane"]:
    la[c] = la[c].fillna(0)
for c in ["TrafDir", "Snow_Priority"]:
    la[c] = blank_to_nan(la[c])
rep = la.geometry.representative_point()
la["lon"], la["lat"] = Transformer.from_crs("EPSG:2263", "EPSG:4326", always_xy=True).transform(rep.x.values, rep.y.values)
la["lion_wkt_2263"] = la.geometry.to_wkt()
seg_table_h = hist_segs[["SegmentID", "lion_key_used", "key_conflict"]].join(la.drop(columns="geometry"))
print(seg_table_h.shape)
```

```python
from difflib import SequenceMatcher

ALIASES = {("7 AVENUE", "ADAM CLAYTON POWELL JR BOULEVARD"), ("8 AVENUE", "FREDERICK DOUGLASS BOULEVARD"),
           ("6 AVENUE", "AVENUE OF THE AMERICAS"), ("F D R DRIVE", "FRANKLIN D ROOSEVELT DRIVE"),
           ("FDR DRIVE", "FRANKLIN D ROOSEVELT DRIVE")}

def same_street(h, l):
    if (h in l) or (l in h) or ((h, l) in ALIASES):
        return True
    return SequenceMatcher(None, h.replace(" ", ""), l.replace(" ", "")).ratio() >= 0.85

rf["name_mismatch2"] = [isinstance(c, str) and c != "" and not same_street(n, c)
                        for c, n in zip(rf["lion_street_n"], rf["road_n"])]
print("Wide rows flagged: before", int(rf["name_mismatch"].sum()), "-> after", int(rf["name_mismatch2"].sum()))

mm2 = rf[rf["name_mismatch2"]].merge(hist_segs[["SegmentID", "lion_key_used"]], on="SegmentID")
print(mm2.groupby("lion_key_used").size())

ex2 = (mm2.merge(hh[key_w + ["Roadway Name"]].drop_duplicates(key_w), on=key_w)
         .assign(year=lambda x: x["Date"].dt.year)
         .groupby(["SegmentID", "Roadway Name", "lion_street_n", "lion_key_used"])
         .agg(rows=("Date", "size"), first=("year", "min"), last=("year", "max"))
         .reset_index().sort_values("rows", ascending=False))
print(len(ex2), "groups")
print(ex2.head(25).to_string())
```

```python
# B. Extended normalization plus alternate names
def norm4(s):
    s = norm3(s)
    for pat, rep in [(r"\bEXPWY\b|\bEXPRWY\b|\bEP\b", "EXPRESSWAY"), (r"\bBX\b", "BRONX"),
                     (r"\bSER RD\b|\bSVC RD\b", "SERVICE ROAD")]:
        s = s.str.replace(pat, rep, regex=True)
    return s.str.replace(r"\s+", " ", regex=True).str.strip()

hh["road_n4"] = norm4(hh["Roadway Name"])
street_n4 = norm4(lion["Street"])
alt_n4 = alt.assign(n=norm4(alt["Street"])).groupby("jid")["n"].agg(set)

hist_segs["lion_street_n4"] = hist_segs["lion_row"].map(
    lambda r: street_n4.loc[int(r)] if pd.notna(r) else np.nan)
hist_segs["lion_alts_n4"] = hist_segs["lion_row"].map(
    lambda r: alt_n4.get(lj.loc[int(r)], set()) if pd.notna(r) else set())

rf4 = (hh[key_w + ["road_n4"]].drop_duplicates(key_w)
         .merge(hist_segs[["SegmentID", "lion_street_n4", "lion_alts_n4"]], on="SegmentID", how="left"))

def ok_row(n, c, alts):
    if not isinstance(c, str) or c == "":
        return True                       # nothing to compare against
    cands = [c] + (list(alts) if isinstance(alts, set) else [])
    return any(same_street(n, x) for x in cands)

rf4["name_mismatch3"] = [not ok_row(n, c, a) for n, c, a in
                         zip(rf4["road_n4"], rf4["lion_street_n4"], rf4["lion_alts_n4"])]
print("Flagged rows: 1607 (first rule) -> 973 (fuzzy+alias) ->", int(rf4["name_mismatch3"].sum()))

# Row-level check by year
print(rf4.assign(year=rf4["Date"].dt.year).groupby("year")["name_mismatch3"]
        .agg(["sum", "size"]).assign(share=lambda x: (x["sum"] / x["size"]).round(3)))

# Residual groups
mm3 = rf4[rf4["name_mismatch3"]].merge(hist_segs[["SegmentID", "lion_key_used"]], on="SegmentID")
ex3 = (mm3.merge(hh[key_w + ["Roadway Name"]].drop_duplicates(key_w), on=key_w)
          .assign(year=lambda x: x["Date"].dt.year)
          .groupby(["SegmentID", "Roadway Name", "lion_street_n4", "lion_key_used"])
          .agg(rows=("Date", "size"), first=("year", "min"), last=("year", "max"))
          .reset_index().sort_values("rows", ascending=False))
print(len(ex3), "groups")
print(ex3.head(25).to_string())
```
