```python
import pandas as pd
import numpy as np
```

```python
df = df.sort_values(by= "Date")
df.head(-10)
```

```python
df["SegmentID"].nunique()
```

```python
h = df.copy()
hour_cols = [f"h{i:02d}" for i in range(24)]
h.columns = list(h.columns[:7]) + hour_cols   # rename by position, not by label
h.columns.tolist()
```

```python

h["Date_raw"] = h["Date"]
h["Date"] = pd.to_datetime(h["Date_raw"], format="%m/%d/%Y", errors="coerce")

print("Unparseable dates:", h["Date"].isna().sum())
print("Range:", h["Date"].min(), "->", h["Date"].max())
print(h["Date"].dt.year.value_counts().sort_index())
print(h["Date"].dt.day_name().value_counts())
```

```python
print(h["Direction"].value_counts(dropna=False))
print(h["Direction"].map(repr).unique())   # reveals trailing spaces or odd casing
```

```python
raw = h[hour_cols]
as_num = raw.apply(lambda s: pd.to_numeric(s.astype(str).str.replace(",", "", regex=False), errors="coerce"))

# Values that were NOT null originally but fail numeric conversion
junk_mask = raw.notna() & as_num.isna()
print("Junk cells per column:\n", junk_mask.sum()[lambda s: s > 0])

# Show the actual offending values
junk_vals = pd.Series(raw.where(junk_mask).stack().astype(str).values)
print(junk_vals.value_counts().head(20))

# Do any values contain commas?
print("Cells with commas:", raw.astype(str).apply(lambda s: s.str.contains(",")).sum().sum())
```

```python
key = ["SegmentID", "Direction", "Date"]
print("Full-row duplicates:", h.duplicated(subset=[c for c in h.columns if c != "Date_raw"]).sum())
print("Duplicate keys:", h.duplicated(key).sum())

dup = h[h.duplicated(key, keep=False)].sort_values(key)
print("Rows involved in duplicate keys:", len(dup))
dup.head(10)
```

```python
h_num = as_num.copy()
h["n_hours"] = h_num.notna().sum(axis=1)
print(h["n_hours"].value_counts().sort_index())

# Where do the partial rows fall? (year, and which hours exist)
partial = h[h["n_hours"] < 24]
print(partial["Date"].dt.year.value_counts().sort_index())
partial[["SegmentID", "Direction", "Date", "n_hours"]].head(15)
```

```python
print(h_num.stack().describe(percentiles=[.01, .05, .5, .95, .99, .999]))
print("Zeros:", (h_num == 0).sum().sum(), "| Negatives:", (h_num < 0).sum().sum())

# Per-hour profile (should show a sensible daily shape)
print(h_num.median().round(0))

# Daily totals: spot suspicious days
daily = h_num.sum(axis=1).where(h["n_hours"] == 24)
print(daily.describe(percentiles=[.001, .01, .5, .99, .999]))
print("All-zero days:", (daily == 0).sum())
```

```python
g = h.groupby("SegmentID")
print("SegmentIDs:", h["SegmentID"].nunique())
print("Segments with >1 Roadway Name:", (g["Roadway Name"].nunique() > 1).sum())
print("Segments with >1 From:", (g["From"].nunique() > 1).sum())
print("Segments with >1 To:", (g["To"].nunique() > 1).sum())

# Examples of segments with multiple names
multi = g["Roadway Name"].nunique()
ex = multi[multi > 1].index[:5]
h[h["SegmentID"].isin(ex)][["SegmentID", "Roadway Name", "From", "To"]].drop_duplicates().sort_values("SegmentID")
```

```python
# What is ID? Relationship between ID and SegmentID
print("Unique IDs:", h["ID"].nunique())
print("IDs mapping to >1 SegmentID:", (h.groupby("ID")["SegmentID"].nunique() > 1).sum())
print("SegmentIDs mapping to >1 ID:", (h.groupby("SegmentID")["ID"].nunique() > 1).sum())
```

```python
cov = (h.groupby("SegmentID")
         .agg(n_dates=("Date", "nunique"),
              n_years=("Date", lambda s: s.dt.year.nunique()),
              first=("Date", "min"),
              last=("Date", "max")))
print(cov[["n_dates", "n_years"]].describe())
print("Segments with only 1 year of data:", (cov["n_years"] == 1).sum(), "of", len(cov))
print(cov["n_dates"].value_counts().sort_index().head(15))
```

```python
# Campaign length: consecutive-day runs per segment and direction
d = (h[["SegmentID", "Direction", "Date"]].drop_duplicates()
       .sort_values(["SegmentID", "Direction", "Date"]))
d["gap"] = d.groupby(["SegmentID", "Direction"])["Date"].diff().dt.days
new_run = d["gap"].isna() | (d["gap"] > 1)
d["campaign"] = new_run.groupby([d["SegmentID"], d["Direction"]]).cumsum()

camp = d.groupby(["SegmentID", "Direction", "campaign"]).agg(days=("Date", "nunique"))
print(camp["days"].value_counts().sort_index())                 # how long is one campaign?
print(camp.groupby(["SegmentID", "Direction"]).size().describe())  # campaigns per segment
```

```python
for c in ["Roadway Name", "From", "To"]:
    s = h[c].astype(str)
    print(f"\n{c}: unique={s.nunique()}, has leading/trailing spaces={(s != s.str.strip()).sum()}, "
          f"not uppercase={(s != s.str.upper()).sum()}")

print(h["From"].value_counts().head(10))
print(h["To"].value_counts().head(10))
print("Non-street markers:",
      h[["From", "To"]].apply(lambda s: s.str.contains("CENSUS|BOUNDARY|END|DEAD", case=False, na=False)).sum().to_dict())
```

```python
for c in hour_cols:
    h[c] = pd.to_numeric(h[c].astype(str).str.replace(",", "", regex=False), errors="coerce")

h_num = h[hour_cols]
print(h[hour_cols].dtypes.value_counts())   # should show int64 or float64
print("NaN count:", h_num.isna().sum().sum())
```

```python
# A. Zero patterns
zero = (h_num == 0)
print("Zeros per hour:\n", zero.sum())
tail0 = (h_num[["h21", "h22", "h23"]] == 0).all(axis=1)
print("Rows with h21-h23 all zero:", tail0.sum())
print(h.loc[tail0, "Date"].dt.year.value_counts().sort_index())
seg_hour_med = h_num.groupby(h["SegmentID"]).transform("median")
print("Zeros where segment-hour median > 100:", (zero & (seg_hour_med > 100)).sum().sum())
```

```python
# B. SegmentID reuse: are different names tied to different years?
def norm(s):
    return (s.str.upper().str.strip()
             .str.replace(r"\bAVE\b", "AVENUE", regex=True)
             .str.replace(r"\bST\b", "STREET", regex=True)
             .str.replace(r"\bRD\b", "ROAD", regex=True)
             .str.replace(r"\bBLVD\b", "BOULEVARD", regex=True))

h["road_n"] = norm(h["Roadway Name"])
n_names = h.groupby("SegmentID")["road_n"].nunique()
print("Segments with >1 normalized name:", (n_names > 1).sum())

x = h[h["SegmentID"].isin(n_names[n_names > 1].index)]
ex = (x.groupby(["SegmentID", "road_n"])["Date"]
        .agg(["min", "max", "count"]).reset_index()
        .sort_values(["SegmentID", "min"]))
ex.head(30)
```

```python
# C. Duplicate pairs: same ID? same names? volume ratio?
key = ["SegmentID", "Direction", "Date"]
d = h[h.duplicated(key, keep=False)].copy()
d["total"] = d[hour_cols].sum(axis=1)
d["rank"] = d.groupby(key).cumcount()
w = d.pivot_table(index=key, columns="rank", values="total")

print("Group sizes:", d.groupby(key).size().value_counts().to_dict())
print("Pairs with different ID:", (d.groupby(key)["ID"].nunique() > 1).sum())
print("Pairs with different name/From/To:",
      (d.groupby(key)[["Roadway Name", "From", "To"]].nunique() > 1).any(axis=1).sum())
print((w[[0, 1]].max(axis=1) / w[[0, 1]].min(axis=1)).describe())
print(d["Date"].dt.year.value_counts().sort_index())
```
